```python
import re

def is_english(text):
    if not isinstance(text, str):
        return False
    return len(text.encode('ascii', 'ignore')) / (len(text) + 1e-5) > 0.85

clean_df = amazon_pairs[
    amazon_pairs['text_customer'].apply(is_english) & 
    amazon_pairs['text_response'].apply(is_english)
].copy()

def clean_tweet(text):
    text = re.sub(r'@[A-Za-z0-9_]+', '', str(text))
    text = re.sub(r'https?://\S+', '', text)
    text = re.sub(r'\^[A-Z0-9]+', '', text)
    return text.strip()

clean_df['clean_customer'] = clean_df['text_customer'].apply(clean_tweet)
clean_df['clean_response'] = clean_df['text_response'].apply(clean_tweet)

clean_df = clean_df[
    (clean_df['clean_customer'].str.len() > 15) & 
    (clean_df['clean_response'].str.len() > 15)
]

clean_df[['clean_customer', 'clean_response']].to_csv('amazon_pairs.csv', index=False)
print(f"Exported {len(clean_df)} clean English pairs to amazon_pairs.csv")
```
